# Exercise 2.A Using built-in functions
## Lab 2. Parsing strings - part codes

```python
acme_part = 'ACME:123-L'
di_part = 'DI:12345-M'
ace_part = 'ACE:1-12'
```

```python

def get_supplier_code(full_string):#Define function to get supplier
    colon = full_string.find(":") # find the index position of the colon
    supplier = full_string[:colon] # for every character before colon 
    return supplier # 

print(get_supplier_code(acme_part))
print(get_supplier_code(di_part))
print(get_supplier_code(ace_part))
```

```python
def get_product_num(full_string): # Define function to get the product code 
    colon = full_string.find(":") # find and store position of :
    hyphen = full_string.find("-") # find and store position of -
    product_code = full_string[colon+1:hyphen] # product num is stored between the colon and hyphen, that's what we need to retrieve
    return product_code

print(get_product_num(acme_part))
print(get_product_num(di_part))
print(get_product_num(ace_part))
```

```python
def get_product_size(full_string): # Define function to get product size
    hyphen = full_string.find("-")
    size = full_string[hyphen+1:]
    return size 

print(get_product_size(acme_part))
print(get_product_size(di_part))
print(get_product_size(ace_part))
```

```python
print(f"For {ace_part}, the supplier is {get_supplier_code(ace_part)}.")
print(f"Part number {get_product_num(ace_part)} is available in size {get_product_size(ace_part)}.")
```
